# Titanic features v1

Survival against one feature at a time, from train.csv. 891 passengers, 342 survived, so the survival rate for all passengers is 0.38. That rate is the grey line marked "all" on every rate chart.

**How to read the charts**

- Left panel: the number of passengers in each group, split into did not survive (0) and survived (1)
- Right panel: the survival rate of each group on a 0 to 1 axis. A rate of 0.74 means 74 of every 100 passengers in the group survived
- n is the number of passengers in the group. One passenger moves the rate of a small group a long way, so read n before the rate

test.csv is not used here because it has no Survived column.

## 1. PassengerId

PassengerId is a row number from 1 to 891. In the top chart each dot is one passenger, placed at 1 if they survived and 0 if they did not. The dots are spread a little above and below 0 and 1 so that they do not sit on top of each other. The bottom chart is the survival rate of each block of 100 IDs.



Survivors and non-survivors are spread evenly along the whole ID range. The rate per block of 100 moves between 0.28 and 0.49 with no trend. In 19 cases out of 20, a block of 100 passengers drawn at random from a population with a rate of 0.38 lands between about 0.29 and 0.48, so this is close to what chance alone produces.

## 5. Ticket type

A ticket is either a number alone (661 of 891 passengers) or a letter prefix followed by a number. Spellings of the same prefix are merged by ignoring capitals, full stops, slashes and spaces, so `A/5`, `A/5.`, `A./5.` and `A.5.` are one type. That leaves 30 types. 4 tickets read `LINE` with no number and are kept as their own type.

The first chart shows the 8 types with 10 or more passengers and folds the other 22 into "Other prefixes". The second chart shows every type.





Number-only tickets and prefixed tickets survive at the same rate: 0.38 against 0.38 for the 230 passengers with any prefix. The prefixes differ from each other. PC is at 0.65 on 60 passengers, while A/5, SOTON/O.Q. and W./C. are all under 0.15. 22 of the 30 types have fewer than 10 passengers, and 10 have only 1.

## 8. Fare by window

Fare is the price of the ticket in GBP. The windows are narrow where fares are bunched and wide where they are sparse: 457 of the 891 passengers paid under 15, and only 53 paid 100 or more. Each window includes its lower bound and excludes its upper bound, so a fare of exactly 10 is in "10 to 15". The 15 passengers with a fare of 0 are their own window.



The survival rate rises with the fare, from 0.07 for the free tickets to 0.74 at 100 and over. The one step down is 25 to 35 at 0.44, after 15 to 25 at 0.46.
